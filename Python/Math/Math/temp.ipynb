```python
import numpy as np
import matplotlib.pyplot as plt

# 定义 x 的范围（x > 0）
x = np.linspace(0.001, 100, 4000)  # 从 arg0 到 arg1，生成 arg2 个点

# 计算 y = log(x)（以自然对数为例）
y = np.log(x)

# 创建图形
plt.figure(figsize=(8, 6))

# 绘制对数曲线
plt.plot(x, y, label='y = log(x)', color='blue')

# 添加标题和标签
plt.title('Logarithmic Curve: y = log(x)')
plt.xlabel('x')
plt.ylabel('y')

# 添加网格
plt.grid(True, linestyle='--', alpha=0.6)

# 显示图例
plt.legend()

# 显示图形
plt.show()
```
